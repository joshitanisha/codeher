# Assignment B4: Machine Learning – Gradient Descent Algorithm

**Problem statement:** Implement Gradient Descent to find the minimum of `f(x) = (x + 3)^2`, starting from `x = 2`.

**Objective:** Understand how Gradient Descent iteratively updates a parameter in the direction opposite to the derivative.

## Theory

Gradient Descent minimizes a function by repeatedly updating its parameter:

$$x_{new} = x_{old} - \alpha f'(x)$$

For the given function:

$$f(x) = (x+3)^2, \qquad f'(x) = 2(x+3)$$

Therefore, the update rule is:

$$x_{new} = x_{old} - \alpha \cdot 2(x+3)$$

The theoretical minimum is at `x = -3`, where `f(x) = 0`.

## 1. Import Required Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Define the Function and Its Derivative

```python
# Cost function
def f(x):
    return (x + 3)**2

# Derivative of the cost function
def derivative(x):
    return 2 * (x + 3)
```

## 3. Implement the Gradient Descent Algorithm

```python
# Initial parameter value
x = 2.0

# Learning rate and number of iterations
learning_rate = 0.1
iterations = 100

# Store values for visualization
x_values = [x]
y_values = [f(x)]

# Gradient Descent
for i in range(iterations):
    grad = derivative(x)                 # Calculate gradient
    x = x - learning_rate * grad         # Update parameter
    x_values.append(x)
    y_values.append(f(x))

print("Final value of x:", x)
print("Minimum value of function:", f(x))
print("Total iterations:", iterations)
```

## 4. Visualize the Function and Minimum Point

```python
# Generate values for plotting the function
x_range = np.linspace(-6, 4, 100)
y_range = f(x_range)

plt.figure(figsize=(8, 5))
plt.plot(x_range, y_range, label='y = (x + 3)^2')
plt.scatter(x, 0, color='red', label=f"Minimum Point ({x}, 0)")
plt.scatter(2, f(2), color='green', label='Starting Point (2, 25)')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Gradient Descent - Function Minimization')
plt.legend()
plt.grid(True)
plt.show()
```

## 6. Result

The Gradient Descent algorithm starts at `x = 2` and moves towards the minimum of the function. With a learning rate of `0.1` and 100 iterations, the result approaches `x = -3` and the function value approaches `0`.

**Theoretical minimum:** `x = -3`, `f(x) = 0`.
